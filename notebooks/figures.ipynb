# TriTower-m6Am Figure Generation

This notebook reproduces the figures from the TriTower-m6Am paper.
Run the cells below to generate ROC/PR curves, ensemble weight pie charts,
CKA heatmaps, and other visualizations.

```python
import os
import sys
import json
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc, precision_recall_curve

# Set up paths
PROJECT_ROOT = os.path.dirname(os.path.abspath(''))
OUTPUT_DIR = os.path.join(PROJECT_ROOT, 'output')
FIGURES_DIR = os.path.join(PROJECT_ROOT, 'figures')
os.makedirs(FIGURES_DIR, exist_ok=True)

# Color palette (colorblind-friendly)
COLORS = {
    'ensemble': '#D62728',
    'rnafm': '#5DADE2',
    'onehot': '#58D68D',
    'rgcn': '#F4D03F',
}

print('Setup complete. Run the following cells to generate figures.')
```

```python
# Load final results
result_path = os.path.join(OUTPUT_DIR, 'final_result.json')
if os.path.exists(result_path):
    with open(result_path) as f:
        results = json.load(f)
    print('Final results loaded.')
    print(f"  Ensemble AUC: {results['final']['auc']:.4f}")
    print(f"  Ensemble MCC: {results['final']['mcc']:.4f}")
    print(f"  Weights: {results['ensemble_weights']}")
else:
    print(f'Result file not found: {result_path}')
    print('Run train.py first to generate results.')
```
